## This file is used to create KS2 data (from KS1)

```python
import pandas as pd
import numpy as np
import os
import random
```

```python
if not os.path.exists("3_KS1_jupyter.csv"):
    exec(open("3_create_ks1.ipynb").read())
else:
    ks1 = pd.read_csv("3_KS1_jupyter.csv")
    ks2 = ks1.copy()
    ks2.columns = ks2.columns.str.replace("KS1", "KS2")
    ks2['KS2_ACADYR'] = "2014/2015"
    ks2['KS2_MATH_new'] = ks2['KS2_MATH'] * np.random.uniform(0.95, 1.05, size=len(ks2))
    ks2['KS2_math_random'] = np.round(np.random.uniform(1, 500, size=len(ks2)))
    ks2['KS2_crazy'] = np.random.uniform(0, 2, size=len(ks2))

    random_sample = np.random.choice(range(1, 501), 1)[0]
    ks2['KS2_MATH_new'] = np.where(ks2['KS2_math_random'] == random_sample,
                                   ks2['KS2_MATH_new'] * ks2['KS2_crazy'],
                                   ks2['KS2_MATH_new'])


    # Manually rescale to the range [0, 100]
    min_value = ks2['KS2_MATH_new'].min()
    max_value = ks2['KS2_MATH_new'].max()
    ks2['KS2_MATH_new'] = ((ks2['KS2_MATH_new'] - min_value) / (max_value - min_value)) * 100

    ks2['KS2_SEN_new'] = np.where(ks2['KS2_MATH'] <= 30, 1, ks2['KS2_SEN'])
    ks2['ks2_sen_random'] = np.round(np.random.uniform(1, 30, size=len(ks2)))
    ks2['KS2_SEN_new'] = np.where(ks2['ks2_sen_random'] == np.random.choice(range(1, 31), 1)[0],
                                  ks2['KS2_SEN_new'] + 1,
                                  ks2['KS2_SEN_new'])
    ks2['KS2_SEN_new'] = np.where(ks2['KS2_SEN_new'] == 2, 0, ks2['KS2_SEN_new'])

    ks2['KS2_MATH'] = ks2['KS2_MATH_new']
    ks2['KS2_SEN'] = ks2['KS2_SEN_new']

    ks2 = ks2.drop(['KS2_MATH_new', 'KS2_math_random', 'KS2_crazy', 'KS2_SEN_new', 'ks2_sen_random'], axis=1)
    ks2.to_csv("5_KS2_jupyter.csv", index=False)
```
